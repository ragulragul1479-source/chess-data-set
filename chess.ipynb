```python

# ============================================================
# CHESS GAME ANALYSIS
# Data Science Project
# Dataset: games.csv
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# 1. Load Dataset
df = pd.read_csv("games.csv")

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
print("\nFirst Five Records:")
print(df.head())

# 2. Display Dataset Columns
print("\nColumn Names:")
print(df.columns.tolist())

# 3. Display Dataset Information
print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe(include="all"))

# 4. Check Missing Values and Duplicates
print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

# 5. Clean Dataset
analysis = df.copy()

# Convert numerical columns
numeric_cols = [
    "turns",
    "white_rating",
    "black_rating",
    "opening_ply",
    "created_at",
    "last_move_at"
]

for col in numeric_cols:
    analysis[col] = pd.to_numeric(
        analysis[col], errors="coerce"
    )

# Remove duplicate records
analysis = analysis.drop_duplicates()

# Remove records missing essential analysis values
analysis = analysis.dropna(
    subset=[
        "winner",
        "turns",
        "white_rating",
        "black_rating"
    ]
)

# Remove invalid game values
analysis = analysis[
    (analysis["turns"] > 0) &
    (analysis["white_rating"] > 0) &
    (analysis["black_rating"] > 0)
].copy()

print("\nData Cleaning Completed!")
print("Cleaned Dataset Shape:", analysis.shape)

# 6. Statistical Analysis
print("\n========== CHESS GAME ANALYSIS ==========")

print("Total Games:", len(analysis))

print(
    "Average Number of Turns:",
    round(analysis["turns"].mean(), 2)
)

print(
    "Maximum Number of Turns:",
    analysis["turns"].max()
)

print(
    "Minimum Number of Turns:",
    analysis["turns"].min()
)

print(
    "Average White Player Rating:",
    round(analysis["white_rating"].mean(), 2)
)

print(
    "Average Black Player Rating:",
    round(analysis["black_rating"].mean(), 2)
)

print(
    "Highest White Rating:",
    analysis["white_rating"].max()
)

print(
    "Highest Black Rating:",
    analysis["black_rating"].max()
)

print(
    "Average Opening Moves:",
    round(analysis["opening_ply"].mean(), 2)
)

# 7. Create Output Folder
os.makedirs("outputs", exist_ok=True)

# 8. Analyze Game Winners
winner_counts = analysis["winner"].value_counts()

print("\n========== GAME WINNERS ==========")
print(winner_counts)

plt.figure(figsize=(8, 5))
winner_counts.plot(kind="bar")

plt.title("Number of Games Won by Each Player")
plt.xlabel("Winner")
plt.ylabel("Number of Games")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.tight_layout()

plt.savefig(
    "outputs/game_winners.png",
    dpi=300
)
plt.show()

# 9. Analyze Victory Status
victory_counts = analysis["victory_status"].value_counts()

print("\n========== VICTORY STATUS ==========")
print(victory_counts)

plt.figure(figsize=(9, 5))
victory_counts.plot(kind="bar")

plt.title("Distribution of Chess Victory Status")
plt.xlabel("Victory Status")
plt.ylabel("Number of Games")
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y")
plt.tight_layout()

plt.savefig(
    "outputs/victory_status.png",
    dpi=300
)
plt.show()

# 10. Analyze Number of Turns
plt.figure(figsize=(10, 5))

plt.hist(
    analysis["turns"],
    bins=30,
    edgecolor="black"
)

plt.title("Distribution of Chess Game Turns")
plt.xlabel("Number of Turns")
plt.ylabel("Frequency")
plt.grid(axis="y")
plt.tight_layout()

plt.savefig(
    "outputs/turns_distribution.png",
    dpi=300
)
plt.show()

# 11. Compare White and Black Player Ratings
plt.figure(figsize=(9, 5))

plt.scatter(
    analysis["white_rating"],
    analysis["black_rating"],
    alpha=0.4
)

plt.title("White Rating vs Black Rating")
plt.xlabel("White Player Rating")
plt.ylabel("Black Player Rating")
plt.grid(True)
plt.tight_layout()

plt.savefig(
    "outputs/player_rating_comparison.png",
    dpi=300
)
plt.show()

# 12. Analyze Top 10 Chess Openings
opening_counts = (
    analysis["opening_name"]
    .dropna()
    .value_counts()
    .head(10)
)

print("\n========== TOP 10 CHESS OPENINGS ==========")
print(opening_counts)

plt.figure(figsize=(12, 6))

opening_counts.sort_values().plot(kind="barh")

plt.title("Top 10 Most Frequently Used Chess Openings")
plt.xlabel("Number of Games")
plt.ylabel("Chess Opening")
plt.grid(axis="x")
plt.tight_layout()

plt.savefig(
    "outputs/top_10_openings.png",
    dpi=300
)
plt.show()

# 13. Compare Average Turns by Winner
turns_by_winner = analysis.groupby("winner")["turns"].mean()

print("\n========== AVERAGE TURNS BY WINNER ==========")
print(turns_by_winner.round(2))

plt.figure(figsize=(8, 5))

turns_by_winner.plot(kind="bar")

plt.title("Average Game Turns by Winner")
plt.xlabel("Winner")
plt.ylabel("Average Number of Turns")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.tight_layout()

plt.savefig(
    "outputs/average_turns_by_winner.png",
    dpi=300
)
plt.show()

# 14. Analyze Rated and Unrated Games
rated_counts = analysis["rated"].value_counts()

print("\n========== RATED AND UNRATED GAMES ==========")
print(rated_counts)

plt.figure(figsize=(7, 5))

rated_counts.plot(kind="bar")

plt.title("Rated vs Unrated Chess Games")
plt.xlabel("Rated Game")
plt.ylabel("Number of Games")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.tight_layout()

plt.savefig(
    "outputs/rated_unrated_games.png",
    dpi=300
)
plt.show()

# 15. Calculate Rating Difference
analysis["Rating_Difference"] = (
    analysis["white_rating"] -
    analysis["black_rating"]
)

print("\n========== RATING DIFFERENCE ANALYSIS ==========")

print(
    "Average Rating Difference:",
    round(analysis["Rating_Difference"].mean(), 2)
)

print(
    "Maximum Rating Difference:",
    analysis["Rating_Difference"].max()
)

print(
    "Minimum Rating Difference:",
    analysis["Rating_Difference"].min()
)

plt.figure(figsize=(10, 5))

plt.hist(
    analysis["Rating_Difference"],
    bins=30,
    edgecolor="black"
)

plt.title("Distribution of Player Rating Differences")
plt.xlabel("White Rating - Black Rating")
plt.ylabel("Frequency")
plt.grid(axis="y")
plt.tight_layout()

plt.savefig(
    "outputs/rating_difference.png",
    dpi=300
)
plt.show()

# 16. Correlation Analysis
correlation = analysis[
    ["turns", "white_rating", "black_rating", "opening_ply"]
].corr()

print("\n========== CORRELATION MATRIX ==========")
print(correlation.round(3))

plt.figure(figsize=(8, 6))

plt.imshow(correlation, aspect="auto")
plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=30,
    ha="right"
)

plt.yticks(
    range(len(correlation.index)),
    correlation.index
)

plt.title("Correlation Between Chess Game Variables")
plt.tight_layout()

plt.savefig(
    "outputs/correlation_matrix.png",
    dpi=300
)
plt.show()

# 17. Save Analysed Dataset
analysis.to_csv(
    "outputs/analysed_chess_games.csv",
    index=False
)

# Save summary reports
winner_counts.to_csv(
    "outputs/winner_summary.csv",
    header=["Number_of_Games"]
)

opening_counts.to_csv(
    "outputs/opening_summary.csv",
    header=["Number_of_Games"]
)

correlation.to_csv(
    "outputs/correlation_summary.csv"
)

# 18. Final Project Summary
print("\n========== PROJECT SUMMARY ==========")

print("Dataset: Chess Games")
print("Total Original Records:", len(df))
print("Total Analysed Records:", len(analysis))
print("Total Columns:", len(analysis.columns))

print("\nWinner Summary:")
print(winner_counts)

print("\nMost Common Chess Opening:")
if not opening_counts.empty:
    print(opening_counts.index[0])
else:
    print("No opening data available.")

print("\nProject Completed Successfully!")
print("All graphs and CSV reports are saved")
print("inside the outputs folder.")
```
